```python
def get_skill_counts(job_df):
    skills = (
        job_df["skills_clean"]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    return skills.value_counts()
```

```python
sde_counts = get_skill_counts(sde_jobs)
data_counts = get_skill_counts(data_jobs)
ai_counts = ai_skills_normalized.value_counts()
```

```python
sde_counts.head(10)
```

```python
sde_skill_pct = (sde_counts / len(sde_jobs) * 100).round(2)

data_counts = get_skill_counts(data_jobs)
data_skill_pct = (data_counts / len(data_jobs) * 100).round(2)
```

```python
sde_salary = sde_jobs[
    sde_jobs["minSalary_clean"].notna() &
    sde_jobs["maxSalary_clean"].notna()
].copy()

data_salary = data_jobs[
    data_jobs["minSalary_clean"].notna() &
    data_jobs["maxSalary_clean"].notna()
].copy()

ai_ml_salary = ai_ml_jobs[
    ai_ml_jobs["minSalary_clean"].notna() &
    ai_ml_jobs["maxSalary_clean"].notna()
].copy()
```

```python
salary_comparison = pd.DataFrame({
    "SDE": [
        sde_salary["minSalary_clean"].median(),
        sde_salary["maxSalary_clean"].median()
    ],
    "Data": [
        data_salary["minSalary_clean"].median(),
        data_salary["maxSalary_clean"].median()
    ],
    "AI_ML": [
        ai_ml_salary["minSalary_clean"].median(),
        ai_ml_salary["maxSalary_clean"].median()
    ]
}, index=["Median Minimum Salary", "Median Maximum Salary"])

salary_comparison
```

```python
pd.Series({
    "SDE": len(sde_salary),
    "Data": len(data_salary),
    "AI_ML": len(ai_ml_salary)
})
```

```python
for job_df in [sde_salary, data_salary, ai_ml_salary]:
    job_df["experience_level"] = pd.cut(
        job_df["minimumExperience"],
        bins=[-1, 0, 2, 5, 10, 100],
        labels=["Fresher", "Junior", "Mid-level", "Senior", "Expert"]
    )
```

```python
salary_by_career_experience = pd.DataFrame({
    "SDE": sde_salary.groupby(
        "experience_level", observed=True
    )["minSalary_clean"].median(),

    "Data": data_salary.groupby(
        "experience_level", observed=True
    )["minSalary_clean"].median(),

    "AI_ML": ai_ml_salary.groupby(
        "experience_level", observed=True
    )["minSalary_clean"].median()
})

salary_by_career_experience
```

```python
career_summary = pd.DataFrame({
    "Career": ["SDE", "Data", "AI/ML"],
    "Job_Count": [
        len(sde_jobs),
        len(data_jobs),
        len(ai_ml_jobs)
    ],
    "Median_Min_Salary": [
        sde_salary["minSalary_clean"].median(),
        data_salary["minSalary_clean"].median(),
        ai_ml_salary["minSalary_clean"].median()
    ],
    "Median_Max_Salary": [
        sde_salary["maxSalary_clean"].median(),
        data_salary["maxSalary_clean"].median(),
        ai_ml_salary["maxSalary_clean"].median()
    ]
})

career_summary
```

```python
skill_comparison = (
    comparison
    .reset_index()
    .rename(columns={"skills_clean": "Skill"})
)

skill_comparison.head(20)
```

```python
experience_salary = (
    salary_by_career_experience
    .reset_index()
    .rename(columns={"experience_level": "Experience_Level"})
)

experience_salary
```

```python
top_skills = skill_comparison[
    (skill_comparison["SDE"] > 0) |
    (skill_comparison["Data"] > 0) |
    (skill_comparison["AI_ML"] > 0)
].copy()

top_skills = top_skills.sort_values(
    "AI_ML",
    ascending=False
).head(50)

top_skills
```

```python
sde_top10 = (
    skill_comparison[["Skill", "SDE"]]
    .sort_values("SDE", ascending=False)
    .head(10)
)

sde_top10
```
